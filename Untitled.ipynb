```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import precision_score, recall_score, make_scorer, accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier, StackingClassifier
from sklearn.feature_selection import mutual_info_regression
from sklearn.naive_bayes import BernoulliNB
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV
```

```python
df = pd.read_csv("../data/dataset/dataset02_filled_missing.csv")
df.head(5)
```

```python
df.dtypes
```

```python
df["customer_tenure"] = df["customer_tenure"] * 1.0
df["subscription_expired"] = df["subscription_expired"] * 1.0
```

```python
if df.isnull().values.any():
    print("Dataset still has missing values. Please address them before proceeding.")
```

```python
X, y = df.drop(columns=["customer_id", "churn"]), df["churn"]
categorical_features = ["plan_tier", "gender"]
```

```python
print(y.sum(), y.count())
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), X.select_dtypes(include=["int64", "float64"]).columns),
        ("categorical", OneHotEncoder(), categorical_features),
    ]
)
```

```python
X_copy = X.copy()
X_copy["gender"] = X_copy["gender"].map({"M": 0, "F": 1})
X_copy["plan_tier"] = X_copy["plan_tier"].map({"Free": 0, "Plus": 1, "Premium": 2})
#X_copy.dtypes
```

```python
discrete_features = X_copy.dtypes == int
def make_mi_scores(X, y, discrete_features):
    mi_scores = mutual_info_regression(X, y, discrete_features=discrete_features)
    mi_scores = pd.Series(mi_scores, name="MI Scores", index=X.columns)
    mi_scores = mi_scores.sort_values(ascending=False)
    return mi_scores

mi_scores = make_mi_scores(X_copy, y, discrete_features)
mi_scores  # show a few features with their MI scores
```

```python
X_copy.drop(columns=["total_orders_60d", "customer_age", "gender", "missing_csat_rate_60d", "converted_rate_60d", "clicked_rate_60d", "total_tickets_60d", "avg_csat_score_all_time", "avg_csat_score_60d", "avg_resolution_hours_60d", "days_since_last_activity", "interaction_60d_share"], inplace=True)
```

```python
pipe = Pipeline([
    ("Preprocessor", preprocessor),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])
```

```python
model = GridSearchCV(
    estimator=pipe,
    param_grid={"knn__n_neighbors": [n for n in range(1, 11)]},
    scoring={'precision': make_scorer(precision_score), 
             'recall': make_scorer(recall_score),
             'accuracy': make_scorer(accuracy_score)},
    refit='precision',
    cv=5
)
```

```python
model.fit(X, y)
```

```python
cv_result = model.cv_results_
plt.figure(figsize=(10, 5))
plt.plot(
    cv_result["mean_test_precision"],
    marker="o",
    label="Mean Precision"
)

plt.plot(
    cv_result["mean_test_recall"],
    marker="o",
    label="Mean Recall"
)

plt.plot(
    cv_result["mean_test_accuracy"],
    marker="o",
    label="Mean Accuracy"
)
plt.xlabel("Parameter combination")
plt.ylabel("Score")
plt.title("Mean Precision vs Mean Recall vs Mean Accuracy for Different Class Weights")
plt.legend()
plt.grid(True)

plt.draw()
```

```python
categorical_features = ["plan_tier"]
preprocessor2 = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), X_copy.select_dtypes(include=["int64", "float64"]).columns),
        ("categorical", OneHotEncoder(), categorical_features),
    ]
)
def min_recall_precision (y_true, y_pred):
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)

    return min(precision, recall)
```

```python
stack = StackingClassifier(
    estimators=[
        (
            "lr",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced"
            )
        ),

        (
            "rf",
            RandomForestClassifier(
                random_state=42,
                class_weight="balanced"
            )
        ),

        (
            "svc",
            CalibratedClassifierCV(
                estimator=SVC(
                    class_weight="balanced"
                ),
                ensemble=False
            )
        )
    ],

    final_estimator=LogisticRegression(
        max_iter=1000
    ),

    cv=5,
    stack_method="predict_proba",
    passthrough=False,
    n_jobs=None
)
pipe2 = Pipeline([
    ("Preprocessor", preprocessor2),
    ("stack", stack)
])


model2 = GridSearchCV(
    estimator=pipe2,
    param_grid={
        "stack__final_estimator__C": [0.1, 1, 10],

        "stack__lr__C": [0.1, 1, 10],
    
        "stack__rf__n_estimators": [150, 200],
        "stack__rf__max_depth": [None],
    
        "stack__svc__estimator__C": [1],
        "stack__svc__estimator__kernel": ["rbf", "linear"]
    },
    scoring={
        "precision": make_scorer(precision_score),
        "recall": make_scorer(recall_score),
        "min_recall_precision": make_scorer(min_recall_precision)
    },
    refit="min_recall_precision",
    cv=4
)
```

```python
model2.fit(X_copy, y)
```

```python
cv_result2 = model2.cv_results_
plt.figure(figsize=(10, 5))
plt.plot(
    cv_result2["mean_test_precision"],
    marker="o",
    label="Mean Precision"
)

plt.plot(
    cv_result2["mean_test_recall"],
    marker="o",
    label="Mean Recall"
)

plt.plot(
    cv_result2["mean_test_min_recall_precision"],
    marker="o",
    label="Mean Min Recall-Precision"
)
plt.xlabel("Parameter combination")
plt.ylabel("Score")
plt.title("Mean Precision vs Mean Recall vs Mean Accuracy for Different Class Weights")
plt.legend()
plt.grid(True)

plt.draw()
```
